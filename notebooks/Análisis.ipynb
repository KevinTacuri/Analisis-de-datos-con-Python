##### 1. Cree un data frame, elimine las columnas ‘name’, ’ticket’, ‘cabin’, ‘boat’, ‘body’ y ‘home.dest’, contabilice la cantidad de valores únicos por columna e impute a la columna ‘age’ mediante la mediana, a la columna ‘embarked’ mediante la moda y elimine ls filas que contenga valores nulos. Además, convierta a las columnas ‘survived’ y ‘pclass’ a tipo object y, cambie las categorías de ‘pclass’ de 1, 2 y 3 a '1st', '2nd', '3rd' y las categorías de ‘survived’ de 0 y 1 a ‘no’ y ‘yes’

```python
import pandas as pd
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df = df.drop(columns=['name', 'ticket', 'cabin', 'boat', 'body', 'home.dest'])
df.head()
```

```python
unicos = df.nunique()
print("Cantidad de valores únicos por columna:\n", unicos)
```

```python
df['age'] = df['age'].fillna(df['age'].median())
df['age'].median()
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df.dropna().head()
```

```python
df['pclass'] = df['pclass'].astype(object)
df['pclass'] = df['pclass'].replace({1: '1st', 2: '2nd', 3: '3rd'})

df['survived'] = df['survived'].astype(object)
df['survived'] = df['survived'].replace({0: 'No', 1: 'Yes'})
```

```python
df.head()
```

##### 2. Cree una columna con fechas aleatorias que vayan desde las 00:00 horas del 1 de enero de 2023 hasta las 23:59 horas del 31 de diciembre de 2023. Luego, cree una columna con el formato para fecha 18 July, 2023 06:12 y otra columna con el formato 18/07/2023 06:12 AM, además, extraiga en columnas diferentes el día, día de semana (en inglés), mes y año.

```python
import datetime
import random

inicio = pd.to_datetime('2023-01-01 00:00:00')
fin = pd.to_datetime('2023-12-31 23:59:59')

def fecha_hora_aleatoria(inicio, fin):
    delta = fin - inicio
    int_delta = int(delta.total_seconds())
    aleatorio_segundos = random.randint(0, int_delta)
    return inicio + pd.to_timedelta(aleatorio_segundos, unit='s')

# Generar la columna de fechas y horas aleatorias
df['date_time'] = [fecha_hora_aleatoria(inicio, fin) for _ in range(len(df))]
df.head()
```

```python
df["date_time_new1"] = df["date_time"].apply(lambda x: x.strftime("%d-%B-%Y %H:%M:%S"))
df["date_time_new2"] = df["date_time"].apply(lambda x: x.strftime("%d/%m/%Y %I:%M:%S %p"))
df.head()
```

```python
df["day"] = df["date_time"].dt.day
df["day_week"] = df["date_time"].dt.day_name()
df["month"] = df["date_time"].dt.month
df["year"] = df["date_time"].dt.year
df.head()
```

##### 3. Realice una normalización de las columnas ‘sibsp’ y ‘parch’ mediante un bucle for, asignándoles los nombres ‘sibsp_norm’ y ‘parch_norm’, respectivamente, y eliminando las columnas originales.

```python
columnas_norm =["sibsp","parch"]

for col in columnas_norm:
    max_val = df[col].max()
    min_val = df[col].min()
    
    df[col] = (df[col] - min_val) / (max_val - min_val)

columnas_resultado = {col: f'{col}_norm' for col in columnas_norm}
df = df.rename(columns=columnas_resultado)

df.head()
```

##### 4. Realice una estandarización de la columna ‘age’ mediante un bucle for, asignándoles el nombre ‘age_std’ y eliminando la columna original. Además, realice una dummización de las columnas ‘pclass’ y ‘embarked’ mediante un bucle for, anteponiendo los nombres ‘pclass_type’ y ‘embarked_type’ a cada categoría, respectivamente, y eliminando las columnas originales. Por último, descargue el último data frame obtenido como un archivo CSV.

```python
#Estandarización de la columna 'age'

columnas_est =["age"]

for col in columnas_est:
    mean = df['age'].mean()
    std = df['age'].std(ddof=0)
    
    df[col] = (df[col] - mean) / std

columnas_resultado = {col: f'{col}_std' for col in columnas_est}

df = df.rename(columns=columnas_resultado)
df.head()
```

```python
#Dummización de la columna 'age'

columnas_dum =["pclass","embarked"]

for col in columnas_dum:
    dummy_variable = pd.get_dummies(df[col])

columnas_resultado = {col: f'{col}_type' for col in columnas_dum}

df = df.rename(columns=columnas_resultado)
df.head()
```
